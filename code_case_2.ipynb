```python
# YÊU CẦU 12: Vẽ bar chart so sánh trung bình các chỉ số giữa 2 bệnh viện

import numpy as np
import matplotlib.pyplot as plt

# Tính giá trị trung bình cho từng chỉ số
trung_binh_a = np.array([
    np_hospital_a[:, 0].mean(),  
    np_hospital_a[:, 1].mean(),  
    np_hospital_a[:, 2].mean(),  
    np_hospital_a[:, 3].mean(),  
    np_hospital_a[:, 4].mean(), 
    np_hospital_a[:, 5].mean()   
])

trung_binh_b = np.array([
    np_hospital_b[:, 0].mean(),
    np_hospital_b[:, 1].mean(),
    np_hospital_b[:, 2].mean(),
    np_hospital_b[:, 3].mean(),
    np_hospital_b[:, 4].mean(),
    np_hospital_b[:, 5].mean()
])

# Tên các chỉ số
ten_chi_so = ['Tuổi', 'Huyết áp', 'Cholesterol', 'Chiều cao', 'Cân nặng', 'BMI']

# Vị trí cột trên trục X
vi_tri = np.arange(len(ten_chi_so))
do_rong = 0.35

# Vẽ biểu đồ cột
plt.figure()
plt.bar(vi_tri - do_rong/2, trung_binh_a, width=do_rong, label='Bệnh viện A')
plt.bar(vi_tri + do_rong/2, trung_binh_b, width=do_rong, label='Bệnh viện B')

# Ghi chú biểu đồ
plt.xticks(vi_tri, ten_chi_so)
plt.title('So sánh giá trị trung bình các chỉ số giữa 2 bệnh viện')
plt.xlabel('Chỉ số')
plt.ylabel('Giá trị trung bình')
plt.legend()

# Hiển thị
plt.show()
```
