**Yêu cầu 7: Broadcasting: chuẩn hoá từng chỉ số (x - mean).** 

- Hospital_a: 

```python

# Chuẩn hóa huyết áp: 
bool_pre_a = arr_blood_pre_a[1:].astype(float) - (arr_blood_pre_a[1:].astype(float)).mean()
np_bool_pre_a = np.array(bool_pre_a)
np.set_printoptions(precision=2, suppress=True)
print('Chuẩn hóa huyết áp: ','\n')
print(np_bool_pre_a)



```

```python
# Chuẩn hóa Cholesterol: 
cholesterol_a = arr_cholesterol_a[1:].astype(float) - (arr_cholesterol_a[1:].astype(float)).mean()
np_cholesterol_a = np.array(cholesterol_a)
np.set_printoptions(precision=2, suppress=True)
print('Chuẩn hóa Cholesterol: ','\n')
print(np_cholesterol_a)
```

```python
# Chuẩn hóa heart_risk: 
heart_risk_a = arr_heart_risk_a[1:].astype(float) - arr_heart_risk_a[1:].astype(float).mean()
np_heart_risk_a = np.array(heart_risk_a)
np.set_printoptions(precision=2, suppress=True)
print('Chuẩn hóa chỉ số Heart_risk: ','\n')
print(np_heart_risk_a)
```
