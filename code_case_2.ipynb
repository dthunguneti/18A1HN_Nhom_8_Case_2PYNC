**Yêu cầu 1: Đọc hospital_A.csv và hospital_B.csv.** 

```python
# Đọc file csv hospital_a.csv : 
hospital_a = []
with open('hospital_a.csv', 'r') as file_a:
    for line in file_a:
        hospital_a.append(line.strip().split(','))



np_hospital_a = np.array(hospital_a)
print('Danh sách file hospital_a: ','\n')
print(np_hospital_a)
```

```python
# Đọc file hospital_b.csv: 
hospital_b = []
with open('hospital_b.csv', 'r') as file_b: 
    for line in file_b: 
        hospital_b.append(line.strip().split(','))
np_hospital_b = np.array(hospital_b)
print('Danh sách file hospital_b: ','\n')
print(np_hospital_b)
```
